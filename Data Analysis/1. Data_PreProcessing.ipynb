# 1. Importing Necessary Libraries

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
```

# 2. Loading the Dataset

```python
raw_df = pd.read_csv("D:\HOPE AI\Capstone Project\Telecom Churn Prediction\Datasets\Raw\WA_Fn-UseC_-Telco-Customer-Churn.csv")
raw_df
```

# 3. Data Preprocessing

```python
raw_df.info()
```

***1.Dropping Customer ID column***

```python
raw_df.drop(columns=["customerID"],inplace=True)
```

***2.Converting wrong categorical column to numeric column***

```python
# Converting Total charges column to numeric column
raw_df['TotalCharges'] = pd.to_numeric(raw_df['TotalCharges'].replace(' ', np.nan), errors='coerce')

# Verify
print("TotalCharges data type:", raw_df['TotalCharges'].dtype)
```

***3.Checking for null values***

```python
raw_df.isnull().sum()
```

```python
# Null value handling
# Replace nan with 0 as they are new customers with tenure = 0
raw_df["TotalCharges"].fillna(0.0,inplace=True)

# Verify
raw_df.isnull().sum()
```

***4. Categorical data handling***

```python
# Applying loop to check the unique values in categorical columns
for col in raw_df.columns:
    if raw_df[col].dtype == "object":
        print(f'{col} : {raw_df[col].unique()}')
```

```python
# cleaning redudant categorical labels
service_cols = [
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
    'TechSupport', 'StreamingTV', 'StreamingMovies']

for col in service_cols:
    raw_df[col] = raw_df[col].replace({"No internet service":"No"})

raw_df['MultipleLines'] = raw_df['MultipleLines'].replace({'No phone service': 'No'})
```

```python
# verif
for col in raw_df.columns:
    if raw_df[col].dtype == "object":
        print(f'{col} : {raw_df[col].unique()}')
```

```python
# Binary Encoding (Map Yes/No and Female/Male to 1/0)
binary_cols = [
    'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 
    'TechSupport', 'StreamingTV', 'StreamingMovies', 
    'PaperlessBilling', 'Churn'
]

for col in binary_cols:
    raw_df[col] = raw_df[col].map({'Yes': 1, 'No': 0})

raw_df['gender'] = raw_df['gender'].map({'Female': 1, 'Male': 0})
```

```python
# One hot encoding for multiclass nominal data
raw_df= pd.get_dummies(
    raw_df, 
    drop_first=True, 
    dtype=int
)
```
