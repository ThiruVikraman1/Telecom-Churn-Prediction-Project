```python
import pandas as pd
import numpy as np
import pickle
```

```python
model = pickle.load(open("Saved Models/Customer_Churn_Prediction_Best_Model.pkl",'rb'))
sc = pickle.load(open("Saved Models/X_scaler.pkl",'rb'))
```

```python
seniorcitizen = int(input("Customer is senior citizen: yes=1,no=0"))
partner = int(input("Customer has partner:yes=1,no=0"))
dependents = int(input("Customer has dependents:yes=1,no=0"))
tenure = int(input("enter tenure in months"))
OnlineSecurity = int(input("Customer has online security:yes=1,no=0"))
TechSupport= int(input("Customer has TechSupport:yes=1,no=0"))
PaperlessBilling = int(input("Customer has paperlessbilling:yes=1,no=0"))
MonthlyCharges = float(input("Enter customer Monthly charges"))
TotalCharges = float(input("Enter customer Total charges"))
InternetService_Fiber_optic = int(input("Customer has fiber optic service:yes=1,no=0"))
InternetService_No = int(input("Customer has internet service:yes=1,no=0"))
Contract_One_year = int(input("Customer contract is 1 year:yes=1,no=0"))
Contract_Two_year = int(input("Customer contract is 2 year:yes=1,no=0"))
PaymentMethod_Credit_card_automatic = int(input("Customer payment is credit card:yes=1,no=0"))
PaymentMethod_Electronic_check = int(input("Customer payment is electronic checkr:yes=1,no=0"))
```

```python
user_input = {
    'SeniorCitizen': seniorcitizen,
    'Partner': partner,
    'Dependents': dependents,
    'tenure': tenure,
    'OnlineSecurity': OnlineSecurity,
    'TechSupport': TechSupport,
    'PaperlessBilling': PaperlessBilling,
    'MonthlyCharges': MonthlyCharges,
    'TotalCharges': TotalCharges,
    'InternetService_Fiber optic': InternetService_Fiber_optic,
    'InternetService_No': InternetService_No,
    'Contract_One year': Contract_One_year,
    'Contract_Two year': Contract_Two_year,
    'PaymentMethod_Credit card (automatic)': PaymentMethod_Credit_card_automatic,
    'PaymentMethod_Electronic check': PaymentMethod_Electronic_check
}
```

```python
user_df = pd.DataFrame([user_input])
```

```python
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
user_df[num_cols] = sc.transform(user_df[num_cols])
```

```python
future_prediction = model.predict(user_df)
print(f"Customer churn prediction = {future_prediction[0]}")
```
