```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
from imblearn.over_sampling import SMOTE
import warnings
warnings.filterwarnings('ignore')
import os
import sys
sys.path.append(os.path.abspath(".."))
sys.path.append(os.path.abspath("../src"))
```

***1. Load the cleaned dataset***

```python
df = pd.read_csv("D:\HOPE AI\Capstone Project\Telecom Churn Prediction\Datasets\Preprocessed\Cleaned_telco_churn.csv")
df.head()
```

***2. Seperating feature and target variables***

```python
independent = df.drop(columns=["Churn"])
dependent = df["Churn"]
```

```python
independent.head()
```

```python
dependent.head()
```

***3. Splitting training and testing data***

```python
X_train, X_test, Y_train, Y_test = train_test_split(independent, dependent, test_size=0.2, random_state=0)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
Y_train.shape
```

```python
Y_test.shape
```

***4. Standard scaling on continuous features***

```python
num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
sc = StandardScaler()

X_train[num_cols] = sc.fit_transform(X_train[num_cols])
X_test[num_cols] = sc.transform(X_test[num_cols])
```

***5. Apply PCA***

```python
from Feature_selection import Feature_Selectors
X_train_pca, X_test_pca, explained_variance = Feature_Selectors.apply_pca(X_train, X_test, 5)
```

```python
print("Selected Training Features Shape:", X_train_pca.shape)
print("Selected Testing Features Shape:", X_test_pca.shape)
```

***6. Churn value Balancing***

```python
# Checking the balance
Y_train.value_counts()
```

```python
# SMOTE for balancing the data
smote = SMOTE(random_state=42)
X_train_k_resampled, Y_train_resampled = smote.fit_resample(X_train_pca, Y_train)
```

```python
# Checking after smote
Y_train_resampled.value_counts()
```

***7. Model Training***

```python
from Models import Classification_Models
from Evaluation_Metrics import ModelEvaluator
```

```python
# Logistic Regression
print("Logistic Regression Model")
y_pred_lr , grid_lr = Classification_Models.logistic(X_train_k_resampled,Y_train_resampled,X_test_pca)
best_lr, cm_lr, report_lr, roc_lr = ModelEvaluator.evaluate(y_pred_lr, grid_lr, X_test_pca, Y_test)

print("Best Parameters:", best_lr)
print("Confusion Matrix:\n", cm_lr)
print("Classification Report:\n", report_lr)
print(f"ROC-AUC Score: {roc_lr:.4f}")
```

```python
# Support vector machine
print("SVM Model")
y_pred_svm, grid_svm = Classification_Models.SVM(X_train_k_resampled, Y_train_resampled, X_test_pca)
best_svm, cm_svm, report_svm, roc_svm = ModelEvaluator.evaluate(y_pred_svm, grid_svm, X_test_pca, Y_test)

print("Best Parameters:", best_svm)
print("Confusion Matrix:\n", cm_svm)
print("Classification Report:\n", report_svm)
print(f"ROC-AUC Score: {roc_svm:.4f}")
```

```python
# K-Nearest neighbour
print("KNN  Model")
y_pred_knn, grid_knn = Classification_Models.KNN(X_train_k_resampled, Y_train_resampled, X_test_pca)
best_knn, cm_knn, report_knn, roc_knn = ModelEvaluator.evaluate(y_pred_knn, grid_knn, X_test_pca, Y_test)

print("Best Parameters:", best_knn)
print("Confusion Matrix:\n", cm_knn)
print("Classification Report:\n", report_knn)
print(f"ROC-AUC Score: {roc_knn:.4f}")
```

```python
# Naive Bayes 
print("Naive Bayes Model")
y_pred_nb, grid_nb = Classification_Models.NaiveBayes(X_train_k_resampled, Y_train_resampled, X_test_pca)
best_nb, cm_nb, report_nb, roc_nb = ModelEvaluator.evaluate(y_pred_nb, grid_nb, X_test_pca, Y_test)

print("Best Parameters:", best_nb)
print("Confusion Matrix:\n", cm_nb)
print("Classification Report:\n", report_nb)
print(f"ROC-AUC Score: {roc_nb:.4f}")
```

```python
#Decision Tree
print("Decision Tree Model")
y_pred_dt, grid_dt = Classification_Models.DecisionTree(X_train_k_resampled, Y_train_resampled, X_test_pca)
best_dt, cm_dt, report_dt, roc_dt = ModelEvaluator.evaluate(y_pred_dt, grid_dt, X_test_pca, Y_test)

print("Best Parameters:", best_dt)
print("Confusion Matrix:\n", cm_dt)
print("Classification Report:\n", report_dt)
print(f"ROC-AUC Score: {roc_dt:.4f}")
```

```python
# Random Forest
print("Random Forest Model")
y_pred_rf, grid_rf = Classification_Models.RandomForest(X_train_k_resampled, Y_train_resampled, X_test_pca)
best_rf, cm_rf, report_rf, roc_rf = ModelEvaluator.evaluate(y_pred_rf, grid_rf, X_test_pca, Y_test)

print("Best Parameters:", best_rf)
print("Confusion Matrix:\n", cm_rf)
print("Classification Report:\n", report_rf)
print(f"ROC-AUC Score: {roc_rf:.4f}")
```
